# Problem Statement 9: Data Visualization II
**Objective:** Plot a box plot for the distribution of age with respect to each gender, along with survival information, using the Titanic dataset.

**Columns Used:** `sex`, `age`, `survived`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
print("Libraries imported!")
```

## Step 1: Load and Inspect the Titanic Dataset

```python
titanic = sns.load_dataset('titanic')

print("Shape:", titanic.shape)
print("\nMissing values in 'age':", titanic['age'].isnull().sum())

# Subset the relevant columns
df = titanic[['sex', 'age', 'survived']].dropna()
print("After dropping NaN rows:", df.shape)

# Map survived to labels
df = df.copy()
df['Survival'] = df['survived'].map({0: 'Did Not Survive', 1: 'Survived'})

df.head(10)
```

## Step 2: Box Plot – Age Distribution by Gender and Survival

```python
plt.figure(figsize=(10, 6))
sns.boxplot(
    x='sex',
    y='age',
    hue='Survival',
    data=df,
    palette={'Survived': 'lightgreen', 'Did Not Survive': 'salmon'},
    width=0.5
)
plt.title('Age Distribution by Gender and Survival Status', fontsize=14)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Age (years)', fontsize=12)
plt.legend(title='Survival Status', fontsize=10)
plt.tight_layout()
plt.show()
```

```python
# Grouped statistics
print("Summary Statistics – Age grouped by Gender and Survival:")
summary = df.groupby(['sex', 'Survival'])['age'].agg(
    Count='count',
    Mean='mean',
    Median='median',
    Std='std',
    Min='min',
    Max='max'
).round(2)
print(summary)
```

## Step 3: Additional Visualizations

```python
# Violin plot for deeper distribution insight
plt.figure(figsize=(10, 6))
sns.violinplot(
    x='sex',
    y='age',
    hue='Survival',
    data=df,
    split=True,
    palette={'Survived': 'lightblue', 'Did Not Survive': 'lightyellow'},
    inner='quartile'
)
plt.title('Violin Plot – Age by Gender and Survival', fontsize=14)
plt.xlabel('Gender')
plt.ylabel('Age')
plt.legend(title='Survival', loc='upper right')
plt.tight_layout()
plt.show()
```

```python
# FacetGrid – Age histogram by sex and survival
g = sns.FacetGrid(df, col='sex', row='Survival', height=3.5, aspect=1.4)
g.map(sns.histplot, 'age', bins=20, color='steelblue', kde=True)
g.set_axis_labels('Age', 'Count')
g.set_titles(col_template='{col_name}', row_template='{row_name}')
g.figure.suptitle('Age Distribution by Gender and Survival', y=1.02, fontsize=13)
plt.tight_layout()
plt.show()
```

```python
# Swarmplot overlay on boxplot
plt.figure(figsize=(10, 6))
sns.boxplot(x='sex', y='age', hue='Survival', data=df,
            palette={'Survived': 'lightgreen', 'Did Not Survive': 'salmon'},
            fliersize=0)
sns.stripplot(x='sex', y='age', hue='Survival', data=df,
              dodge=True, alpha=0.3, size=3, palette='dark:#888')

handles, labels = plt.gca().get_legend_handles_labels()
plt.legend(handles[:2], labels[:2], title='Survival')
plt.title('Box + Strip Plot – Age by Gender and Survival')
plt.xlabel('Gender')
plt.ylabel('Age')
plt.tight_layout()
plt.show()
```

## Step 4: Observations and Inferences

Based on the box plots and statistics above:

1. **Females who survived** had a median age of ~28 years, while females who did not survive had a slightly higher median age.

2. **Males who survived** were slightly younger on average than males who did not survive. Younger males had a better chance of survival.

3. **Females overall** had a higher survival rate than males, regardless of age group — supporting the "women and children first" evacuation protocol.

4. **Children (young ages)** show up in both male and female survival groups, suggesting children were prioritized in evacuation.

5. **Older males** (age > 50) had a very low survival rate compared to other groups.

6. The **IQR (box size)** for females is narrower than males, indicating less age variability among surviving females.
