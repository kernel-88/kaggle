```python
df.dtypes
```

## Filling the missing values

```python
# Filling blank Age column values with medians
df['Age'].fillna(df['Age'].median(), inplace=True)

# Filling the Embarked column blank values with mode (most frequent)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
```

```python
df.info()
```

```python
# Dropping the Cabin column since it has too many missing value rows (around 77 %)
df.drop(columns = ['Cabin'], inplace=True)
```

```python
df.isnull().sum()
```

```python
### Normalizing the numeric columns with min max scaling
scaler = MinMaxScaler()
# Selecting the numeric data variables
numeric_cols = ['Age', 'SibSp', 'Parch', 'Fare']
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

print("Normalized numeric columns")
df[numeric_cols].head()
```

### Categorical Variables to Quantitative conversion

```python
le = LabelEncoder()
df['Sex_Encoded'] = le.fit_transform(df['Sex'])
df['Embarked_Encoded'] = le.fit_transform(df['Embarked'])
print('Original vs Label Encoded')
print(df[['Sex', 'Sex_Encoded', 'Embarked', 'Embarked_Encoded']].head())
```
