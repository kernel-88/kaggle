# Problem Statement 8: Data Visualization I
**Objective:**
1. Use the Titanic inbuilt Seaborn dataset to find patterns.
2. Plot a histogram for the distribution of ticket fares.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot style
sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (10, 6)

print("Libraries imported!")
```

## Step 1: Load the Titanic Dataset

```python
titanic = sns.load_dataset('titanic')

print("Dataset Shape:", titanic.shape)
print("\nColumn Names:", titanic.columns.tolist())
print("\nFirst 5 rows:")
titanic.head()
```

```python
print("Basic Statistics:")
titanic.describe()
```

```python
print("Missing values:")
print(titanic.isnull().sum())
```

## Step 2: Pattern Analysis with Seaborn

```python
# Survival count by gender
plt.figure(figsize=(7, 5))
sns.countplot(x='sex', hue='survived', data=titanic, palette='Set2')
plt.title('Survival Count by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()

print("\nObservation: Females had a much higher survival rate than males.")
```

```python
# Survival by passenger class
plt.figure(figsize=(7, 5))
sns.countplot(x='pclass', hue='survived', data=titanic, palette='Set1')
plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class (1=First, 2=Second, 3=Third)')
plt.ylabel('Count')
plt.legend(title='Survived', labels=['No', 'Yes'])
plt.tight_layout()
plt.show()

print("\nObservation: 1st class passengers had the highest survival rate.")
```

```python
# Survival rate by class
plt.figure(figsize=(7, 5))
sns.barplot(x='pclass', y='survived', data=titanic, palette='Blues_d')
plt.title('Survival Rate by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Survival Rate')
plt.tight_layout()
plt.show()
```

```python
# Age distribution by survival
plt.figure(figsize=(9, 5))
sns.kdeplot(data=titanic, x='age', hue='survived', fill=True, palette='coolwarm', alpha=0.5)
plt.title('Age Distribution by Survival Status')
plt.xlabel('Age')
plt.ylabel('Density')
plt.tight_layout()
plt.show()

print("\nObservation: Children (age < 10) had higher survival rates.")
```

## Step 3: Histogram of Fare Distribution

```python
# Histogram of fare
plt.figure(figsize=(10, 5))
plt.hist(titanic['fare'].dropna(), bins=40, color='steelblue', edgecolor='black', alpha=0.8)
plt.axvline(titanic['fare'].mean(), color='red', linestyle='--', label=f"Mean: {titanic['fare'].mean():.2f}")
plt.axvline(titanic['fare'].median(), color='green', linestyle='--', label=f"Median: {titanic['fare'].median():.2f}")
plt.title('Distribution of Ticket Fares')
plt.xlabel('Fare')
plt.ylabel('Number of Passengers')
plt.legend()
plt.tight_layout()
plt.show()

print(f"\nFare Statistics:")
print(f"  Mean   : {titanic['fare'].mean():.2f}")
print(f"  Median : {titanic['fare'].median():.2f}")
print(f"  Std Dev: {titanic['fare'].std():.2f}")
print(f"  Min    : {titanic['fare'].min():.2f}")
print(f"  Max    : {titanic['fare'].max():.2f}")
print("\nObservation: Fare is heavily right-skewed; most passengers paid low fares but a few paid very high fares.")
```

```python
# Seaborn histplot with KDE
plt.figure(figsize=(10, 5))
sns.histplot(titanic['fare'].dropna(), bins=40, kde=True, color='coral')
plt.title('Fare Distribution with KDE Curve')
plt.xlabel('Fare ($)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Fare by Pclass – Violin Plot
plt.figure(figsize=(9, 5))
sns.violinplot(x='pclass', y='fare', data=titanic, palette='muted', inner='quartile')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.tight_layout()
plt.show()

print("\nObservation: 1st class passengers paid significantly higher fares with a much wider spread.")
```
