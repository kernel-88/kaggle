# Problem Statement 3: Descriptive Statistics – Measures of Central Tendency and Variability
**Objective:** Compute summary statistics grouped by categorical variables and analyze the Iris dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

## Part 1: Summary Statistics Grouped by Categorical Variable
We use the Titanic dataset. Categorical variable = `Pclass`, Numeric variable = `Fare`.

```python
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
titanic = pd.read_csv(url)

print("Dataset Shape:", titanic.shape)
titanic.head()
```

```python
# Summary statistics of Fare grouped by Pclass
grouped_stats = titanic.groupby('Pclass')['Fare'].agg(
    Mean='mean',
    Median='median',
    Min='min',
    Max='max',
    Std='std',
    Count='count'
).round(2)

print("Summary Statistics of Fare grouped by Passenger Class:")
grouped_stats
```

```python
# Create a list of fare values per class
fare_by_class = {
    pclass: titanic[titanic['Pclass'] == pclass]['Fare'].dropna().tolist()
    for pclass in titanic['Pclass'].unique()
}

for pclass, fares in sorted(fare_by_class.items()):
    print(f"\nClass {pclass}: First 5 fares = {fares[:5]}")
```

```python
# Visualize Fare distribution by Pclass
plt.figure(figsize=(8, 5))
sns.boxplot(x='Pclass', y='Fare', data=titanic, palette='Set2')
plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

## Part 2: Statistical Details for Iris Dataset Species
We load the Iris dataset and compute statistics for each species.

```python
from sklearn.datasets import load_iris

iris_data = load_iris()
iris = pd.DataFrame(iris_data.data, columns=iris_data.feature_names)
iris['species'] = pd.Categorical.from_codes(iris_data.target, iris_data.target_names)

print("Iris Dataset Shape:", iris.shape)
iris.head()
```

```python
# Statistical details per species
for species in iris['species'].unique():
    print(f"\n{'='*60}")
    print(f"Species: {species.upper()}")
    print('='*60)
    subset = iris[iris['species'] == species]
    stats = subset.describe(percentiles=[0.25, 0.50, 0.75]).round(3)
    print(stats)
```

```python
# Compute custom statistics
print("\nCustom Statistics by Species:")
custom_stats = iris.groupby('species').agg(['mean', 'median', 'std', 'min', 'max']).round(3)
custom_stats
```

```python
# Percentile values per species
print("\nPercentile Analysis (Sepal Length):")
for species in iris['species'].unique():
    data = iris[iris['species'] == species]['sepal length (cm)']
    print(f"\n{species}:")
    for p in [25, 50, 75, 90]:
        print(f"  {p}th percentile = {np.percentile(data, p):.3f}")
```

```python
# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x='species', y='sepal length (cm)', data=iris, palette='Set1', ax=axes[0])
axes[0].set_title('Sepal Length by Species')

sns.barplot(x='species', y='petal length (cm)', data=iris, palette='Set2', ci='sd', ax=axes[1])
axes[1].set_title('Mean Petal Length by Species (with SD)')

plt.tight_layout()
plt.show()
```
